```python
# === Cell 1: Grounding + Data Snapshot (no plots) ===

import pandas as pd
import numpy as np

# --- CONFIG: update this path if needed ---
EXCEL_PATH = "MS_Capstone_Final_Quality_Check_Summary.xlsx"
SHEET_NAME = "clean_data"      # <- required sheet name

# --- Read the sheet without touching your source file ---
df = pd.read_excel(EXCEL_PATH, sheet_name=SHEET_NAME, dtype=str)  # read as str to avoid implicit coercions
df.columns = [c.strip() for c in df.columns]

# --- Helper: safe converters for dates & numerics (in-memory only, no file changes) ---
def to_dt(s):
    return pd.to_datetime(s, errors="coerce", dayfirst=True, utc=False)

def to_num(s):
    # Allow commas, % signs; keep NaN if empty
    if s is None:
        return np.nan
    if isinstance(s, (int, float, np.number)):
        return s
    s2 = str(s).replace(",", "").replace("%", "").strip()
    return pd.to_numeric(s2, errors="coerce")

# --- Identify common columns if present ---
col = {k:k for k in df.columns}  # simple alias map
get = lambda name: name if name in df.columns else None

# Typical columns in your clean layer
C_TICKER     = get("Ticker")
C_SECTOR     = get("Sector")
C_CAP        = get("Cap Category")
C_RESULT     = get("Result Date")
C_T          = get("Closing Price (T)")
C_T1         = get("Closing Price (T+1)")
C_T2         = get("Closing Price (T+2)")
C_T3         = get("Closing Price (T+3)")
C_TDATE      = get("_T_date")
C_T1DATE     = get("_T1_date")
C_T2DATE     = get("_T2_date")
C_T3DATE     = get("_T3_date")
C_QA         = get("qa_flags")  # may or may not exist

# Sector-specific / fundamentals (existence-checked)
C_EPS        = get("EPS in Rs")
C_OPM        = get("OPM %")
C_NETPROFIT  = get("Net Profit")
C_SALES      = get("Sales")
C_REVENUE    = get("Revenue")
C_FINMARG    = get("Financing Margin %")
C_GNPA       = get("Gross NPA %")
C_NNPA       = get("Net NPA %")

# Derived returns (if present)
C_RET3       = get("3-Day Return (%)")
C_AVG3       = get("3-Day Avg Price Post Result")

# --- Create an analysis copy with typed columns (no write-back) ---
dfa = df.copy()

# Parse dates where columns exist
for dc in [C_RESULT, C_TDATE, C_T1DATE, C_T2DATE, C_T3DATE]:
    if dc:
        dfa[dc + "__dt"] = to_dt(dfa[dc])

# Parse numeric columns commonly used
for nc in [C_T, C_T1, C_T2, C_T3, C_EPS, C_OPM, C_NETPROFIT, C_SALES, C_REVENUE,
           C_FINMARG, C_GNPA, C_NNPA, C_RET3, C_AVG3]:
    if nc:
        dfa[nc + "__num"] = dfa[nc].map(to_num)

# --- Snapshot metrics ---
print("=== DATA SNAPSHOT ===")
print(f"Rows: {len(dfa):,}")
if C_TICKER: print(f"Unique Tickers: {dfa[C_TICKER].nunique(dropna=True)}")
if C_SECTOR:
    sectors = sorted([s for s in dfa[C_SECTOR].dropna().unique()])
    print(f"Sectors ({len(sectors)}): {', '.join(sectors)}")
if C_CAP:
    caps = sorted([s for s in dfa[C_CAP].dropna().unique()])
    print(f"Cap Categories ({len(caps)}): {', '.join(caps)}")

# Date coverage
if C_RESULT:
    dmin = dfa[C_RESULT + "__dt"].min()
    dmax = dfa[C_RESULT + "__dt"].max()
    print(f"Result Date coverage: {dmin.date() if pd.notna(dmin) else 'NA'}  →  {dmax.date() if pd.notna(dmax) else 'NA'}")

# Events per ticker (top 10)
if C_TICKER:
    print("\nTop 10 tickers by event count:")
    print(dfa[C_TICKER].value_counts().head(10))

# --- Missingness check on key columns (only those present) ---
key_cols = [C_RESULT, C_TDATE, C_T, C_T1DATE, C_T1, C_T2DATE, C_T2, C_T3DATE, C_T3,
            C_EPS, C_OPM, C_NETPROFIT, C_SALES, C_REVENUE, C_FINMARG, C_GNPA, C_NNPA]
key_cols = [c for c in key_cols if c]  # drop Nones

print("\n=== Missingness on key columns ===")
miss = []
for c in key_cols:
    m = dfa[c].isna().sum()
    miss.append((c, int(m), round(100*m/len(dfa), 2)))
miss_df = pd.DataFrame(miss, columns=["Column", "Missing", "Missing_%"]).sort_values("Missing_%", ascending=False)
display(miss_df)

## --- QA flags summary (if available) ---
#if C_QA:
#    # split by comma and stack
#    exploded = (
#        dfa[[C_QA]]
#        .assign(_tmp=lambda x: x[C_QA].fillna("").str.split(","))
#        .explode("_tmp")
#    )
#    exploded["_tmp"] = exploded["_tmp"].str.strip()
#    qa_counts = (exploded.loc[exploded["_tmp"] != "", "_tmp"].value_counts().rename_axis("qa_flag").reset_index(name="count"))
#    print("\n=== QA flags summary ===")
#    display(qa_counts)
#else:
#    print("\n(qa_flags column not found; skipping QA summary.)")
#
# --- Quick dtype overview (analysis copy) ---
print("\n=== Dtype overview (analysis copy) ===")
dtype_overview = pd.DataFrame({
    "column": dfa.columns,
    "dtype": [str(t) for t in dfa.dtypes]
})
display(dtype_overview)
```

```python
# We DO NOT modify original columns in-place; we only create temporary datetime copies for checks.
def _to_dt(s):
    return pd.to_datetime(s, errors="coerce")

# Temporary datetime views (leave df unchanged)
dt_cols = {}
for c in ["Result Date", "_T_date", "_T1_date", "_T2_date", "_T3_date"]:
    if c in df.columns:
        dt_cols[c] = _to_dt(df[c])

print("\n--- Basic coverage checks (recomputed) ---")
print("Rows:", len(df))
if "Ticker" in df.columns:
    print("Unique tickers:", df["Ticker"].nunique())
if "Sector" in df.columns:
    print("Unique sectors:", df["Sector"].nunique())

# Missing counts on key price/date fields (using original df)
keys = ["Result Date", "_T_date", "Closing Price (T)", "_T1_date", "Closing Price (T+1)",
        "_T2_date", "Closing Price (T+2)", "_T3_date", "Closing Price (T+3)"]
present_keys = [k for k in keys if k in df.columns]
miss_summary = df[present_keys].isna().sum().sort_values(ascending=False).rename("missing_count").to_frame()
display(miss_summary)

# --- Recomputed QA flags (aligns with new logic discussed) ---
flags = []

# 1) T_DATE_AFTER_RESULT: _T_date strictly AFTER result date
if {"Result Date","_T_date"}.issubset(dt_cols.keys()):
    mask_after = dt_cols["_T_date"].notna() & dt_cols["Result Date"].notna() & (dt_cols["_T_date"] > dt_cols["Result Date"])
    cnt_after = int(mask_after.sum())
    flags.append(("T_DATE_AFTER_RESULT", cnt_after))
    print(f"T_DATE_AFTER_RESULT: {cnt_after}")
    if cnt_after:
        display(df.loc[mask_after, ["Ticker", "Result Date", "_T_date"]].head(10))

# 2) NON_MONOTONIC_T_DATES: (T <= T+1 <= T+2 <= T+3) among the dates that exist
def non_monotonic_row(i):
    seq = []
    for c in ["_T_date", "_T1_date", "_T2_date", "_T3_date"]:
        if c in dt_cols and pd.notna(dt_cols[c].iat[i]):
            seq.append(dt_cols[c].iat[i])
    return any(seq[j] > seq[j+1] for j in range(len(seq)-1))

if set(["_T_date","_T1_date","_T2_date","_T3_date"]).issubset(dt_cols.keys()):
    mask_nonmono = pd.Series([non_monotonic_row(i) for i in range(len(df))], index=df.index)
    cnt_nonmono = int(mask_nonmono.sum())
    flags.append(("NON_MONOTONIC_T_DATES", cnt_nonmono))
    print(f"NON_MONOTONIC_T_DATES: {cnt_nonmono}")
    if cnt_nonmono:
        display(df.loc[mask_nonmono, ["Ticker","Result Date","_T_date","_T1_date","_T2_date","_T3_date"]].head(10))

# 3) OUT_OF_RANGE_Financing_Margin_%: expected [0, 100]
if "Financing Margin %" in df.columns:
    fm = pd.to_numeric(df["Financing Margin %"], errors="coerce")
    mask_fm = fm.notna() & ((fm < 0) | (fm > 100))
    cnt_fm = int(mask_fm.sum())
    flags.append(("OUT_OF_RANGE_Financing_Margin_%", cnt_fm))
    print(f"OUT_OF_RANGE_Financing_Margin_%: {cnt_fm}")
    if cnt_fm:
        display(df.loc[mask_fm, ["Ticker","Result Date","Financing Margin %"]].head(10))

# 4) OUT_OF_RANGE_Tax_%: expected [0, 100]
if "Tax %" in df.columns:
    tax = pd.to_numeric(df["Tax %"], errors="coerce")
    mask_tax = tax.notna() & ((tax < 0) | (tax > 100))
    cnt_tax = int(mask_tax.sum())
    flags.append(("OUT_OF_RANGE_Tax_%", cnt_tax))
    print(f"OUT_OF_RANGE_Tax_%: {cnt_tax}")
    if cnt_tax:
        display(df.loc[mask_tax, ["Ticker","Result Date","Tax %"]].head(10))

# 5) OUT_OF_RANGE_OPM_%: expected [0, 100] (supports either 'OPM %' or 'OPM%')
opm_col = "OPM %" if "OPM %" in df.columns else ("OPM%" if "OPM%" in df.columns else None)
if opm_col:
    opm = pd.to_numeric(df[opm_col], errors="coerce")
    mask_opm = opm.notna() & ((opm < 0) | (opm > 100))
    cnt_opm = int(mask_opm.sum())
    flags.append(("OUT_OF_RANGE_OPM_%", cnt_opm))
    print(f"OUT_OF_RANGE_OPM_%: {cnt_opm}  (column: {opm_col})")
    if cnt_opm:
        display(df.loc[mask_opm, ["Ticker","Result Date", opm_col]].head(10))

# 6) Price sanity: warn on negatives (should be none)
for pc in ["Closing Price (T)", "Closing Price (T+1)", "Closing Price (T+2)", "Closing Price (T+3)"]:
    if pc in df.columns:
        p = pd.to_numeric(df[pc], errors="coerce")
        cnt_neg = int(((p.notna()) & (p < 0)).sum())
        if cnt_neg:
            print(f"WARNING: {pc} has {cnt_neg} negative values.")

# 7) Duplicate keys
if {"Ticker","Result Date"}.issubset(df.columns):
    dups = df.duplicated(subset=["Ticker","Result Date"], keep=False)
    cnt_dups = int(dups.sum())
    print(f"\nDuplicate (Ticker, Result Date) rows: {cnt_dups}")
    if cnt_dups:
        display(df.loc[dups, ["Ticker","Result Date","_T_date","_T1_date","_T2_date","_T3_date"]].head(20))

print("\n--- Recomputed flag counts (cross-check) ---")
flag_df = pd.DataFrame(flags, columns=["qa_flag","count"]).sort_values("count", ascending=False)
display(flag_df.reset_index(drop=True))
```

```python
# === EDA setup (non-destructive): make a working copy and add helper columns ===
import numpy as np
import pandas as pd

# Safety: ensure df exists
assert 'df' in globals(), "The DataFrame `df` must be loaded in earlier cells."

eda = df.copy(deep=True)

# 1) Temporary datetime views (do not overwrite original columns)
to_dt = lambda s: pd.to_datetime(s, errors="coerce")
eda_dt = {
    "Result Date": to_dt(eda.get("Result Date")),
    "_T_date":     to_dt(eda.get("_T_date")),
    "_T1_date":    to_dt(eda.get("_T1_date")),
    "_T2_date":    to_dt(eda.get("_T2_date")),
    "_T3_date":    to_dt(eda.get("_T3_date")),
}

# 2) Numeric views for price columns (safe coercion)
for pc in ["Closing Price (T)", "Closing Price (T+1)", "Closing Price (T+2)", "Closing Price (T+3)"]:
    if pc in eda.columns:
        eda[f"eda_num_{pc}"] = pd.to_numeric(eda[pc], errors="coerce")

# 3) Compute 1-day, 2-day, 3-day post-result prices & returns (temporary)
P0 = eda.get("eda_num_Closing Price (T)")
P1 = eda.get("eda_num_Closing Price (T+1)")
P2 = eda.get("eda_num_Closing Price (T+2)")
P3 = eda.get("eda_num_Closing Price (T+3)")

# Average of next 3 trading days (require at least 2 valid days)
eda["eda_post3_avg_price"] = np.where(
    pd.concat([P1.notna(), P2.notna(), P3.notna()], axis=1).sum(axis=1) >= 2,
    pd.concat([P1, P2, P3], axis=1).mean(axis=1, skipna=True),
    np.nan
)

# Percentage returns
def pct_ret(start, end):
    return np.where((start.notna()) & (start != 0) & end.notna(), (end - start) / start * 100.0, np.nan)

eda["eda_ret_T_to_T1_pct"] = pct_ret(P0, P1)
eda["eda_ret_T_to_T3avg_pct"] = pct_ret(P0, eda["eda_post3_avg_price"])

# 4) Optional movement label (threshold = 0.5%: Up/Flat/Down)
THRESH = 0.5
ret3 = eda["eda_ret_T_to_T3avg_pct"]
eda["eda_movement_label"] = np.where(
    ret3.notna(),
    np.where(ret3 > THRESH, "Up", np.where(ret3 < -THRESH, "Down", "Flat")),
    np.nan
)

print("EDA setup complete.")
print("Working columns added: ['eda_post3_avg_price','eda_ret_T_to_T1_pct','eda_ret_T_to_T3avg_pct','eda_movement_label']")
print("Base df is unchanged.")
```

```python
# ===== Cell N — EDA: Price & label sanity (uses existing `clean_df`) =====

# --- Safety check: make sure clean_df exists in memory ---
if "df" not in globals():
    raise RuntimeError("`clean_df` is not defined. Please run your 'Health checks: setup' cell first.")

df = df.copy()  # do NOT mutate the original

# --- Helper: safe datetime parsing (won't change underlying data) ---
def _dt(s):
    return pd.to_datetime(s, errors="coerce")

# Pull date series (as parsed views)
dT   = _dt(df.get("_T_date"))
dT1  = _dt(df.get("_T1_date"))
dT2  = _dt(df.get("_T2_date"))
dT3  = _dt(df.get("_T3_date"))
dRes = _dt(df.get("Result Date"))

# --- Calendar-day gaps (NaN if either side missing) ---
gap_cal_T1 = (dT1 - dT).dt.days
gap_cal_T2 = (dT2 - dT1).dt.days
gap_cal_T3 = (dT3 - dT2).dt.days

# --- Business-day gaps (Mon–Fri; holiday-agnostic, NaN if missing) ---
def _bus_gap(a, b):
    out = np.full(len(df), np.nan, dtype="float64")
    mask = a.notna() & b.notna()
    if mask.any():
        # np.busday_count counts business days in [start, end); for "next trading day" this equals 1
        out[mask] = np.busday_count(a[mask].values.astype("datetime64[D]"),
                                    b[mask].values.astype("datetime64[D]"),
                                    weekmask="1111100")  # Mon-Fri
    return pd.Series(out, index=df.index)

gap_bus_T1 = _bus_gap(dT, dT1)
gap_bus_T2 = _bus_gap(dT1, dT2)
gap_bus_T3 = _bus_gap(dT2, dT3)

# --- Package gaps for quick inspection ---
gaps_cal_df = pd.DataFrame({
    "T→T+1 (calendar days)": gap_cal_T1,
    "T+1→T+2 (calendar days)": gap_cal_T2,
    "T+2→T+3 (calendar days)": gap_cal_T3,
})
gaps_bus_df = pd.DataFrame({
    "T→T+1 (business days)": gap_bus_T1,
    "T+1→T+2 (business days)": gap_bus_T2,
    "T+2→T+3 (business days)": gap_bus_T3,
})

print("Calendar-day gaps — non-null counts:\n", gaps_cal_df.notna().sum(), "\n")
print("Business-day gaps — non-null counts:\n", gaps_bus_df.notna().sum(), "\n")

print("Calendar-day gap distributions (value counts):")
for c in gaps_cal_df.columns:
    vc = gaps_cal_df[c].value_counts(dropna=True).sort_index()
    print(f"\n{c}\n{vc.to_string()}")

print("\nBusiness-day gap distributions (value counts):")
for c in gaps_bus_df.columns:
    vc = gaps_bus_df[c].value_counts(dropna=True).sort_index()
    print(f"\n{c}\n{vc.to_string()}")

# --- 3-Day Return (%) (use existing if present; else compute ad hoc) ---
# Try to find an existing "3-Day Return (%)" column, else compute from prices
def _find_three_day_return_col(cols):
    candidates = [c for c in cols if ("return" in c.lower() and "3" in c) or ("3-day" in c.lower() and "%" in c)]
    # Prefer exact / clean match if present
    exact = [c for c in candidates if c.strip().lower() in {"3-day return (%)", "3-day avg return (%)"}]
    return (exact[0] if exact else (candidates[0] if candidates else None))

ret_col = _find_three_day_return_col(df.columns)

if ret_col is None:
    # Compute ad hoc from prices: ((avg(T+1..T+3) - T) / T) * 100
    pT  = pd.to_numeric(df.get("Closing Price (T)"), errors="coerce")
    pT1 = pd.to_numeric(df.get("Closing Price (T+1)"), errors="coerce")
    pT2 = pd.to_numeric(df.get("Closing Price (T+2)"), errors="coerce")
    pT3 = pd.to_numeric(df.get("Closing Price (T+3)"), errors="coerce")
    avg_post = pd.concat([pT1, pT2, pT3], axis=1).mean(axis=1)
    three_day_return = ((avg_post - pT) / pT) * 100.0
    ret_series = three_day_return.rename("3-Day Return (%) [computed]")
    print("\nNote: No explicit 3-Day Return (%) column found; computed ad hoc from price columns.")
else:
    ret_series = pd.to_numeric(df[ret_col], errors="coerce").rename(ret_col)
    print(f"\nUsing existing return column: {ret_col}")

# Summary stats for returns
ret_dropna = ret_series.dropna()
ret_stats = pd.DataFrame({
    "count": [ret_dropna.shape[0]],
    "mean":  [ret_dropna.mean()],
    "std":   [ret_dropna.std()],
    "min":   [ret_dropna.min()],
    "p25":   [ret_dropna.quantile(0.25)],
    "p50":   [ret_dropna.median()],
    "p75":   [ret_dropna.quantile(0.75)],
    "max":   [ret_dropna.max()],
    "skew":  [ret_dropna.skew()],
}).T
ret_stats.columns = ["3-Day Return (%) — summary"]
print("\n3-Day Return (%) — summary statistics:\n", ret_stats.round(4), "\n")

# --- Movement label balance ---
def _find_label_col(cols):
    cands = [c for c in cols if "movement" in c.lower() and "label" in c.lower()]
    return cands[0] if cands else None

label_col = _find_label_col(df.columns)
if label_col:
    label_vc = df[label_col].astype("category").value_counts(dropna=False)
    total = int(label_vc.sum())
    print(f"Movement Label column: {label_col}")
    print("Class balance:\n", label_vc.to_string(), "\n")
    # Simple imbalance metric: max class share
    max_share = (label_vc.max() / total) if total else np.nan
    print(f"Imbalance (max class share): {max_share:.3f}")
else:
    print("No Movement Label column found — skipping class balance.\n")

# --- Optional visuals ---
SHOW_PLOTS = True  # set False if running headless

if SHOW_PLOTS:
    # Histogram of 3-Day Return (%)
    plt.figure(figsize=(7,4.5))
    ret_dropna.plot(kind="hist", bins=30, edgecolor="black")
    plt.axvline(0, linestyle="--")
    # reference threshold lines at ±0.5% if you use that rule
    plt.axvline(0.5, linestyle=":")
    plt.axvline(-0.5, linestyle=":")
    plt.title("3-Day Return (%) Distribution")
    plt.xlabel("3-Day Return (%)")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

    # Bar charts for gap distributions (calendar & business) — T→T+1 only for compactness
    for title, series in [
        ("T→T+1 (calendar days)", gaps_cal_df["T→T+1 (calendar days)"].dropna()),
        ("T→T+1 (business days)", gaps_bus_df["T→T+1 (business days)"].dropna()),
    ]:
        vc = series.value_counts().sort_index()
        plt.figure(figsize=(6,3.8))
        vc.plot(kind="bar")
        plt.title(f"Gap distribution: {title}")
        plt.xlabel("Gap")
        plt.ylabel("Count")
        plt.tight_layout()
        plt.show()
```

```python
# ===== Cell — EDA: Feature distributions & Up/Down contrasts (uses existing `df`) =====
from scipy import stats

# ---------- 0) Safety: pick the working DataFrame ----------
if "df" in globals() and isinstance(df, pd.DataFrame):
    eda_df = df.copy()
elif "clean_df" in globals() and isinstance(clean_df, pd.DataFrame):
    eda_df = clean_df.copy()
else:
    raise RuntimeError("No DataFrame found. Please ensure your clean sheet is loaded into `df` (or `clean_df`).")

# ---------- 1) Helpers ----------
def _num(s):
    return pd.to_numeric(s, errors="coerce")

def _coalesce(*cols):
    """Return the first non-null per-row among provided Series (aligned)."""
    out = None
    for c in cols:
        s = eda_df[c] if isinstance(c, str) else c
        if out is None:
            out = s
        else:
            out = out.where(out.notna(), s)
    return out

def _find_return_col(cols):
    cands = [c for c in cols if "return" in c.lower() and "3" in c]
    # prefer exact name if present
    exact = [c for c in cands if c.strip().lower() in {"3-day return (%)", "3 day return (%)"}]
    return (exact[0] if exact else (cands[0] if cands else None))

# build / find 3-day return (%)
ret_col = _find_return_col(eda_df.columns)
if ret_col is None:
    pT  = _num(eda_df.get("Closing Price (T)"))
    pT1 = _num(eda_df.get("Closing Price (T+1)"))
    pT2 = _num(eda_df.get("Closing Price (T+2)"))
    pT3 = _num(eda_df.get("Closing Price (T+3)"))
    avg_post = pd.concat([pT1, pT2, pT3], axis=1).mean(axis=1)
    three_day_ret = ((avg_post - pT) / pT) * 100.0
    ret_name = "3-Day Return (%) [computed]"
    print("No explicit '3-Day Return (%)' found; computed from price columns.")
else:
    three_day_ret = _num(eda_df[ret_col])
    ret_name = ret_col

# unified Topline: Revenue if available else Sales
topline = _coalesce("Revenue", "Sales")
topline_name = "Topline (Revenue/Sales)"

# core features list (existing names in your file)
features = [
    (ret_name, three_day_ret),
    ("EPS in Rs", _num(eda_df.get("EPS in Rs"))),
    ("OPM %", _num(eda_df.get("OPM %"))),
    ("Net Profit", _num(eda_df.get("Net Profit"))),
    (topline_name, _num(topline)),
]

sector_col = "Sector" if "Sector" in eda_df.columns else None
cap_col    = "Cap Category" if "Cap Category" in eda_df.columns else None

# movement label (target)
label_col = None
for c in eda_df.columns:
    if "movement" in c.lower() and "label" in c.lower():
        label_col = c
        break

# ---------- 2) Plotting helpers (matplotlib only) ----------
SHOW_PLOTS = True  # set to False to skip figures

def plot_hist(series, title, xlabel):
    s = series.dropna()
    if s.empty: 
        print(f"[plot] Skipped '{title}' — no data.")
        return
    plt.figure(figsize=(7,4.2))
    plt.hist(s, bins=30, edgecolor="black")
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

def plot_box_by(series, by_col, by_title):
    if by_col is None or by_col not in eda_df.columns:
        print(f"[plot] Skipped boxplot by {by_title} — column not found.")
        return
    tmp = pd.DataFrame({"val": series, "by": eda_df[by_col]}).dropna()
    if tmp.empty:
        print(f"[plot] Skipped boxplot by {by_title} — no data.")
        return
    # Matplotlib boxplot by groups
    groups = [g["val"].values for _, g in tmp.groupby("by")]
    labels = [str(k) for k, _ in tmp.groupby("by")]
    plt.figure(figsize=(max(6, 1.2*len(labels)), 4.2))
    plt.boxplot(groups, labels=labels, showfliers=True)
    plt.title(f"{series.name} — by {by_title}")
    plt.ylabel(series.name)
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()

# ---------- 3) Up vs Down contrasts ----------
def cohens_d(x, y):
    x = np.asarray(x, float)
    y = np.asarray(y, float)
    x = x[~np.isnan(x)]
    y = y[~np.isnan(y)]
    if x.size < 2 or y.size < 2:
        return np.nan
    nx, ny = x.size, y.size
    sx2, sy2 = x.var(ddof=1), y.var(ddof=1)
    # pooled std (unbiased)
    s_p = np.sqrt(((nx - 1)*sx2 + (ny - 1)*sy2) / (nx + ny - 2))
    if s_p == 0:
        return 0.0
    return (x.mean() - y.mean()) / s_p

def cliffs_delta(x, y):
    """Brute-force Cliff's delta (fine for sample sizes here)."""
    x = np.asarray(x, float)
    y = np.asarray(y, float)
    x = x[~np.isnan(x)]
    y = y[~np.isnan(y)]
    if x.size == 0 or y.size == 0:
        return np.nan
    gt = 0
    lt = 0
    for xi in x:
        gt += np.sum(xi > y)
        lt += np.sum(xi < y)
    nxy = x.size * y.size
    return (gt - lt) / nxy

def ttest_welch(x, y):
    try:
        res = stats.ttest_ind(x, y, equal_var=False, nan_policy="omit")
        return float(res.pvalue)
    except Exception:
        return np.nan

def mannwhitney_p(x, y):
    try:
        # Use two-sided; requires at least one observation in each group
        res = stats.mannwhitneyu(x, y, alternative="two-sided")
        return float(res.pvalue)
    except Exception:
        return np.nan

# detect Up/Down labels
if label_col is None:
    print("No Movement Label column found — skipping Up vs Down contrasts.\n")
    contrasts_df = pd.DataFrame()
else:
    # normalize labels to 'Up'/'Down' if possible (don’t mutate df)
    lab = eda_df[label_col].astype(str).str.strip().str.lower()
    up_mask = lab.isin({"up", "rise", "positive", "1"})
    down_mask = lab.isin({"down", "fall", "negative", "-1", "0"})  # include '0' if your rule maps neutral to Down; adjust if needed

    # if neither mapping worked, fall back to value_counts top-2 categories
    if up_mask.sum() == 0 and down_mask.sum() == 0:
        vc = eda_df[label_col].astype("category").value_counts()
        if len(vc) >= 2:
            top_vals = vc.index[:2].tolist()
            up_mask = eda_df[label_col] == top_vals[0]
            down_mask = eda_df[label_col] == top_vals[1]

    rows = []
    for fname, series in features:
        s = _num(series)
        x_up = s[up_mask].dropna().values
        x_dn = s[down_mask].dropna().values
        n_up, n_dn = x_up.size, x_dn.size
        mean_up, mean_dn = (np.nan, np.nan)
        med_up, med_dn = (np.nan, np.nan)
        diff_mean, diff_med = (np.nan, np.nan)
        t_p, mw_p, d_eff, cd_eff = (np.nan, np.nan, np.nan, np.nan)

        if n_up > 0: 
            mean_up, med_up = x_up.mean(), np.median(x_up)
        if n_dn > 0: 
            mean_dn, med_dn = x_dn.mean(), np.median(x_dn)
        if n_up > 0 and n_dn > 0:
            diff_mean = mean_up - mean_dn
            diff_med = med_up - med_dn
            d_eff = cohens_d(x_up, x_dn)
            cd_eff = cliffs_delta(x_up, x_dn)
            t_p = ttest_welch(x_up, x_dn)
            mw_p = mannwhitney_p(x_up, x_dn)

        rows.append({
            "Feature": fname,
            "n_Up": n_up, "n_Down": n_dn,
            "mean_Up": mean_up, "mean_Down": mean_dn, "Δmean (Up-Down)": diff_mean,
            "median_Up": med_up, "median_Down": med_dn, "Δmedian (Up-Down)": diff_med,
            "t-test p": t_p, "Mann–Whitney p": mw_p,
            "Cohen's d": d_eff, "Cliff's δ": cd_eff
        })

    contrasts_df = pd.DataFrame(rows)
    # pretty print
    print("\n=== Up vs Down: Summary contrasts ===")
    with pd.option_context('display.max_rows', None, 'display.float_format', '{:,.4f}'.format):
        display(contrasts_df)

# ---------- 4) Distribution plots ----------
if SHOW_PLOTS:
    for name, series in features:
        series = pd.Series(series, name=name)
        plot_hist(series, title=f"{name} — Overall distribution", xlabel=name)
        plot_box_by(series, sector_col, "Sector")
        plot_box_by(series, cap_col, "Cap Category")

# ---------- 5) Quick numeric summaries by Sector / Cap ----------
def summarize_by(group_col, series, label):
    if group_col is None or group_col not in eda_df.columns:
        return None
    tmp = pd.DataFrame({label: _num(series), "grp": eda_df[group_col]}).dropna()
    if tmp.empty:
        return None
    return tmp.groupby("grp")[label].agg(["count","mean","std","median","min","max"]).sort_index()

print("\n=== Group summaries (mean/median etc.) ===")
for name, series in features:
    sec_sum = summarize_by(sector_col, series, name)
    if sec_sum is not None:
        print(f"\n{name} — by Sector")
        display(sec_sum.round(4))
    cap_sum = summarize_by(cap_col, series, name)
    if cap_sum is not None:
        print(f"\n{name} — by Cap Category")
        display(cap_sum.round(4))
```

```python
# ===== Cell — EDA: Correlations & importance previews (clean_data headers) =====
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Optional deps (graceful fallbacks)
try:
    from sklearn.feature_selection import mutual_info_classif
    SKLEARN_OK = True
except Exception:
    SKLEARN_OK = False
    print("Note: scikit-learn not found — skipping mutual information.\n")

try:
    import statsmodels.api as sm
    from statsmodels.stats.outliers_influence import variance_inflation_factor
    STATSMODELS_OK = True
except Exception:
    STATSMODELS_OK = False
    print("Note: statsmodels not found — skipping VIF calculation.\n")

# ---------- 0) Safety: use the DataFrame already loaded ----------
if "df" not in globals() or not isinstance(df, pd.DataFrame):
    raise RuntimeError("Expected clean data in variable `df` from the earlier cell.")
base = df.copy()

# ---------- 1) Columns (exact names from clean_data) ----------
RET_COL = "3-Day Return (%)"
AVG_POST_COL = "3-Day Avg Price Post Result"
LABEL_COL = "Movement Label (1,0)"

PRICE_COLS = [
    "Closing Price (T)", "Closing Price (T+1)", "Closing Price (T+2)", "Closing Price (T+3)"
]
DATE_COLS = ["_T_date", "_T1_date", "_T2_date", "_T3_date"]

# Fundamentals present in your clean_data header
FUND_CANDIDATES = [
    "EPS in Rs", "OPM %", "Net Profit", "Profit before tax", "Tax %",
    "Other Income", "Interest", "Depreciation", "Sales", "Revenue",
    "Financing Profit", "Financing Margin %"
]
FUND_COLS = [c for c in FUND_CANDIDATES if c in base.columns]

def to_num(series):
    return pd.to_numeric(series, errors="coerce")

# ---------- 2) Build numeric matrix ----------
# Keep fundamentals + return (for correlation views)
num_df = base[FUND_COLS].apply(to_num)
if RET_COL in base.columns:
    num_df = pd.concat([num_df, to_num(base[RET_COL]).rename(RET_COL)], axis=1)
else:
    raise RuntimeError(f"Expected '{RET_COL}' in the sheet.")

# Drop all-NaN or constant columns to keep matrices readable
keep_cols = []
for c in num_df.columns:
    s = num_df[c].dropna()
    if s.empty:
        continue
    if s.nunique() <= 1:
        continue
    keep_cols.append(c)
num_df = num_df[keep_cols]

print("Numeric features considered (fundamentals + return):")
print(", ".join(num_df.columns.tolist()), "\n")

# ---------- 3) Correlation matrices ----------
pear = num_df.corr(method="pearson")
spear = num_df.corr(method="spearman")

def corr_heatmap(df_corr, title):
    if df_corr.empty:
        print(f"[heatmap] Skipped '{title}' — empty correlation matrix.\n")
        return
    plt.figure(figsize=(max(7, 0.55*df_corr.shape[1]), max(6, 0.55*df_corr.shape[0])))
    im = plt.imshow(df_corr.values, aspect='auto')
    plt.colorbar(im, fraction=0.046, pad=0.04)
    plt.title(title)
    plt.xticks(range(df_corr.shape[1]), df_corr.columns, rotation=60, ha='right', fontsize=8)
    plt.yticks(range(df_corr.shape[0]), df_corr.index, fontsize=8)
    for i in range(df_corr.shape[0]):
        for j in range(df_corr.shape[1]):
            val = df_corr.values[i, j]
            if np.isfinite(val):
                plt.text(j, i, f"{val:.2f}", ha="center", va="center", fontsize=7)
    plt.tight_layout()
    plt.show()

print("Top Pearson correlations with", RET_COL)
display(pear[RET_COL].dropna().sort_values(ascending=False))

print("\nTop Spearman correlations with", RET_COL)
display(spear[RET_COL].dropna().sort_values(ascending=False))

corr_heatmap(pear.round(2), f"Pearson correlation heatmap (fundamentals + {RET_COL})")
corr_heatmap(spear.round(2), f"Spearman correlation heatmap (fundamentals + {RET_COL})")

# ---------- 4) Point-biserial (fundamentals vs Movement Label (1,0)) ----------
if LABEL_COL in base.columns:
    y = pd.to_numeric(base[LABEL_COL], errors="coerce")  # expects 0/1 already
    if y.notna().sum() >= 10:
        def point_biserial(x, y):
            m = (~x.isna()) & (~y.isna())
            if m.sum() < 3:
                return np.nan
            return np.corrcoef(x[m].astype(float), y[m].astype(float))[0, 1]

        pb_rows = []
        for c in num_df.columns:
            if c == RET_COL:
                continue  # skip the return when assessing fundamentals vs label
            r = point_biserial(num_df[c], y)
            pb_rows.append({"Feature": c, "Point-biserial r (with Movement Label)": r})
        pb_df = pd.DataFrame(pb_rows).sort_values(
            by="Point-biserial r (with Movement Label)", key=lambda s: s.abs(), ascending=False
        )
        print("\nPoint-biserial correlations (|r| sorted) — fundamentals vs Movement Label (1,0):")
        display(pb_df)
    else:
        print("Movement Label has insufficient non-missing rows for point-biserial.\n")
else:
    print(f"'{LABEL_COL}' not found — skipping label-based correlations.\n")

# ---------- 5) Mutual information (non-linear signal preview) ----------
if SKLEARN_OK and (LABEL_COL in base.columns):
    y = pd.to_numeric(base[LABEL_COL], errors="coerce")
    if y.notna().sum() >= 10:
        # Use fundamentals only (avoid leakage from return/prices)
        X = base[FUND_COLS].apply(to_num)
        m = y.notna()
        X = X.loc[m].copy()
        yv = y.loc[m].astype(int).values
        X_imp = X.fillna(X.median(numeric_only=True))
        try:
            mi = mutual_info_classif(X_imp.values, yv, discrete_features=False, random_state=42)
            mi_df = pd.DataFrame({"Feature": X.columns, "Mutual Information": mi}).sort_values(
                "Mutual Information", ascending=False
            )
            print("\nMutual information with Movement Label (higher = more non-linear predictive signal):")
            display(mi_df)
        except Exception as e:
            print(f"Mutual information computation failed: {e}\n")
    else:
        print("Movement Label has insufficient non-missing rows for mutual information.\n")
elif not SKLEARN_OK:
    pass  # notice already printed above

# ---------- 6) Multicollinearity flags ----------
# (a) Highly correlated fundamental pairs (Pearson |r| ≥ 0.85)
if FUND_COLS:
    fund_df = base[FUND_COLS].apply(to_num)
    # drop all-NaN/constant
    valid = []
    for c in fund_df.columns:
        s = fund_df[c].dropna()
        if s.empty: 
            continue
        if s.nunique() <= 1:
            continue
        valid.append(c)
    fund_df = fund_df[valid]
    if not fund_df.empty:
        fund_pear = fund_df.corr(method="pearson").abs()
        HI_R = 0.85
        pairs = []
        cols = fund_pear.columns
        for i in range(len(cols)):
            for j in range(i+1, len(cols)):
                r = float(fund_pear.iloc[i, j])
                if np.isfinite(r) and r >= HI_R:
                    pairs.append((cols[i], cols[j], r))
        if pairs:
            print(f"\nHighly correlated fundamental pairs (|Pearson r| ≥ {HI_R}):")
            for a, b, r in sorted(pairs, key=lambda t: t[2], reverse=True):
                print(f"  - {a} <> {b}: r = {r:.2f}")
        else:
            print("\nNo fundamental pairs exceed the high-correlation threshold.")
    else:
        print("\nNo valid fundamental columns for collinearity check.")
else:
    print("\nNo fundamental columns detected — skipping collinearity check.")

# (b) Optional VIF on fundamentals (median-imputed), if statsmodels present
if STATSMODELS_OK and FUND_COLS:
    X_vif = base[FUND_COLS].apply(to_num)
    X_vif = X_vif.fillna(X_vif.median(numeric_only=True))
    # drop any remaining non-finite columns
    keep = [c for c in X_vif.columns if np.isfinite(X_vif[c]).all()]
    X_vif = X_vif[keep]
    if X_vif.shape[1] >= 2:
        Xc = sm.add_constant(X_vif.values)
        vif_vals = [variance_inflation_factor(Xc, i+1) for i in range(X_vif.shape[1])]
        vif_df = pd.DataFrame({"Feature": X_vif.columns, "VIF": vif_vals}).sort_values("VIF", ascending=False)
        print("\nVariance Inflation Factors for fundamentals (rule-of-thumb: VIF > 5~10 indicates concern):")
        display(vif_df)
    else:
        print("VIF skipped — not enough fundamental features after cleaning.\n")
elif not STATSMODELS_OK:
    pass  # notice already printed above
```

```python
# ===== Cell — EDA: Time / Seasonality views (clean_data schema) =====

if "df" not in globals() or not isinstance(df, pd.DataFrame):
    raise RuntimeError("Expected clean data in variable `df` from the earlier cell.")

work = df.copy()

# --- Helpers ---
def parse_period_series(per_s: pd.Series, result_s: pd.Series) -> pd.Series:
    """
    Return a pandas datetime (month-end) for each row representing the fiscal quarter 'Period'.
    1) Try to parse Period directly (e.g., 'Jun 2024', or datetime-like)
    2) If missing, derive from Result Date using Indian FY quarters (Q-MAR)
    Output is normalized to month-end and used for ordering on the x-axis.
    """
    # Try direct parse: strings like "Jun 2024"
    p1 = pd.to_datetime(per_s, format="%b %Y", errors="coerce")

    # Generic parse for datetime-like text (e.g., '2024-06-01 00:00:00')
    p2 = pd.to_datetime(per_s, errors="coerce")

    # Prefer p1 when available, else p2
    p = p1.combine_first(p2)

    # Where still NA, derive from Result Date
    rd = pd.to_datetime(result_s, errors="coerce")
    # Use Indian FY quarters with Q-MAR (FY ends in March)
    q_end = rd.dt.to_period("Q-MAR").dt.end_time
    p = p.combine_first(q_end)

    # Normalize to month-end (helps consistent ordering/labels)
    p = (pd.to_datetime(p, errors="coerce") + pd.offsets.MonthEnd(0)).dt.normalize()
    return p

# --- Build period axis (month-end timestamps) and labels ---
work["_Period_dt"] = parse_period_series(work.get("Period"), work.get("Result Date"))
work["_Period_lbl"] = work["_Period_dt"].dt.strftime("%b %Y")

# --- Numeric fields we need ---
RET = "3-Day Return (%)"
EPS = "EPS in Rs"
OPM = "OPM %"

work[RET] = pd.to_numeric(work[RET], errors="coerce")
work[EPS] = pd.to_numeric(work[EPS], errors="coerce")
work[OPM] = pd.to_numeric(work[OPM], errors="coerce")

# ---------- 1) Median 3-Day Return (%) by quarter over time ----------
ret_med = (
    work[[ "_Period_dt", "_Period_lbl", RET ]]
    .dropna(subset=["_Period_dt", RET])
    .groupby(["_Period_dt", "_Period_lbl"], as_index=False)[RET].median()
    .sort_values("_Period_dt")
)

print("Median 3-Day Return (%) by quarter (first 10 rows):")
display(ret_med.head(10))

plt.figure(figsize=(10, 4))
plt.plot(ret_med["_Period_dt"].values, ret_med[RET].values, marker="o")
plt.title("Median 3-Day Return (%) by Quarter (time-ordered)")
plt.xlabel("Quarter (month-end)")
plt.ylabel("Median 3-Day Return (%)")
# Tidy x-axis labels
xt = ret_med["_Period_lbl"].values
step = max(1, len(xt)//10)
plt.xticks(ret_med["_Period_dt"].values[::step], xt[::step], rotation=45, ha="right")
plt.tight_layout()
plt.show()

# ---------- 2) Median EPS by quarter over time ----------
eps_med = (
    work[[ "_Period_dt", "_Period_lbl", EPS ]]
    .dropna(subset=["_Period_dt", EPS])
    .groupby(["_Period_dt", "_Period_lbl"], as_index=False)[EPS].median()
    .sort_values("_Period_dt")
)

print("Median EPS by quarter (first 10 rows):")
display(eps_med.head(10))

plt.figure(figsize=(10, 4))
plt.plot(eps_med["_Period_dt"].values, eps_med[EPS].values, marker="o")
plt.title("Median EPS by Quarter (time-ordered)")
plt.xlabel("Quarter (month-end)")
plt.ylabel("Median EPS (Rs)")
xt = eps_med["_Period_lbl"].values
step = max(1, len(xt)//10)
plt.xticks(eps_med["_Period_dt"].values[::step], xt[::step], rotation=45, ha="right")
plt.tight_layout()
plt.show()

# ---------- 3) Median OPM % by quarter over time ----------
opm_med = (
    work[[ "_Period_dt", "_Period_lbl", OPM ]]
    .dropna(subset=["_Period_dt", OPM])
    .groupby(["_Period_dt", "_Period_lbl"], as_index=False)[OPM].median()
    .sort_values("_Period_dt")
)

print("Median OPM % by quarter (first 10 rows):")
display(opm_med.head(10))

plt.figure(figsize=(10, 4))
plt.plot(opm_med["_Period_dt"].values, opm_med[OPM].values, marker="o")
plt.title("Median OPM % by Quarter (time-ordered)")
plt.xlabel("Quarter (month-end)")
plt.ylabel("Median OPM (%)")
xt = opm_med["_Period_lbl"].values
step = max(1, len(xt)//10)
plt.xticks(opm_med["_Period_dt"].values[::step], xt[::step], rotation=45, ha="right")
plt.tight_layout()
plt.show()

# ---------- 4) Seasonal boxplots of returns by fiscal quarter ----------
# Fiscal quarter buckets from month of _Period_dt with Q-MAR convention:
# Q1=Apr–Jun (labeled "Q1 (Jun)"), Q2=Jul–Sep, Q3=Oct–Dec, Q4=Jan–Mar
month = work["_Period_dt"].dt.month
fq = pd.Series(index=work.index, dtype="object")

fq[(month>=4) & (month<=6)]  = "Q1 (Jun)"
fq[(month>=7) & (month<=9)]  = "Q2 (Sep)"
fq[(month>=10) & (month<=12)] = "Q3 (Dec)"
fq[(month>=1) & (month<=3)]  = "Q4 (Mar)"
work["_Fiscal_Quarter"] = fq

# Prepare data for boxplot
bp = work[[RET, "_Fiscal_Quarter"]].dropna()
order = ["Q1 (Jun)", "Q2 (Sep)", "Q3 (Dec)", "Q4 (Mar)"]
bp["_Fiscal_Quarter"] = pd.Categorical(bp["_Fiscal_Quarter"], categories=order, ordered=True)
bp = bp.sort_values("_Fiscal_Quarter")

# Small summary table
print("\nSeasonal medians of 3-Day Return (%) by fiscal quarter:")
display(bp.groupby("_Fiscal_Quarter")[RET].median().to_frame("Median 3-Day Return (%)"))

# Draw boxplot (matplotlib)
plt.figure(figsize=(7, 4))
groups = [bp.loc[bp["_Fiscal_Quarter"]==lab, RET].values for lab in order]
plt.boxplot(groups, labels=order, showmeans=True)
plt.title("3-Day Return (%) by Fiscal Quarter (Q-MAR)")
plt.xlabel("Fiscal Quarter")
plt.ylabel("3-Day Return (%)")
plt.tight_layout()
plt.show()
```

```python
# ===== Cell — EDA: Cross-plots (relationships) [FIXED LOWESS LENGTH] =====
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

assert "df" in globals(), "Please run the earlier cells that load `df` first."

# Exact headers
COL_RET = "3-Day Return (%)"
COL_EPS = "EPS in Rs"
COL_OPM = "OPM %"
COL_NP  = "Net Profit"
COL_SALES = "Sales"
COL_REV   = "Revenue"
COL_SECTOR = "Sector"
COL_MOV = "Movement Label (1,0)"
COL_FM = "Financing Margin %"

work = df.copy()

# Coerce numerics
for c in [COL_RET, COL_EPS, COL_OPM, COL_NP, COL_SALES, COL_REV, COL_FM, COL_MOV]:
    if c in work.columns:
        work[c] = pd.to_numeric(work[c], errors="coerce")

def lowess_smooth(x, y, frac=0.25):
    """
    Returns (x_sorted, y_smooth) with SAME LENGTHS.
    Tries statsmodels.lowess; otherwise uses centered rolling mean of
    window ~= frac * n (odd), which preserves length.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    m = np.isfinite(x) & np.isfinite(y)
    x, y = x[m], y[m]
    if len(x) < 5:
        order = np.argsort(x)
        return x[order], y[order]

    order = np.argsort(x)
    xs, ys = x[order], y[order]

    # Prefer statsmodels LOWESS if available
    try:
        from statsmodels.nonparametric.smoothers_lowess import lowess as sm_lowess
        sm = sm_lowess(ys, xs, frac=max(0.05, min(0.9, frac)), return_sorted=True)
        return sm[:, 0], sm[:, 1]
    except Exception:
        # Fallback: centered rolling mean with odd window; preserves length == len(xs)
        n = len(xs)
        w = max(5, int(n * frac))
        if w % 2 == 0:
            w += 1  # ensure odd window
        y_sm = pd.Series(ys).rolling(window=w, center=True, min_periods=1).mean().to_numpy()
        return xs, y_sm

# ---------------- 1) EPS vs 3-Day Return (%) ----------------
sub = work[[COL_EPS, COL_RET, COL_MOV]].dropna(subset=[COL_EPS, COL_RET])
mask_up = (sub[COL_MOV] == 1)

plt.figure(figsize=(7, 5))
plt.scatter(sub.loc[mask_up, COL_EPS], sub.loc[mask_up, COL_RET], s=18, alpha=0.6, label="Up")
plt.scatter(sub.loc[~mask_up & ~sub[COL_MOV].isna(), COL_EPS],
            sub.loc[~mask_up & ~sub[COL_MOV].isna(), COL_RET], s=18, alpha=0.6, label="Down")

xs, ys = lowess_smooth(sub[COL_EPS], sub[COL_RET], frac=0.3)
if len(xs) > 1:
    plt.plot(xs, ys, linewidth=2, label="LOWESS trend")

plt.title("EPS vs 3-Day Return (%)")
plt.xlabel("EPS in Rs")
plt.ylabel("3-Day Return (%)")
plt.legend()
plt.tight_layout()
plt.show()

# ---------------- 2) OPM % vs Net Profit — scaled by sector ----------------
if COL_SECTOR in work.columns:
    z = work[[COL_SECTOR, COL_NP, COL_OPM]].dropna(subset=[COL_SECTOR, COL_NP, COL_OPM]).copy()
    def robust_scale(group):
        med = group.median()
        iqr = group.quantile(0.75) - group.quantile(0.25)
        if not (np.isfinite(iqr) and iqr > 0):
            iqr = group.std(ddof=0) or 1.0
        return (group - med) / iqr

    if len(z):
        z["NP_scaled_by_sector"] = z.groupby(COL_SECTOR)[COL_NP].transform(robust_scale)
        for sec, g in z.groupby(COL_SECTOR):
            plt.figure(figsize=(6, 4))
            plt.scatter(g["NP_scaled_by_sector"], g[COL_OPM], s=20, alpha=0.7)
            xs, ys = lowess_smooth(g["NP_scaled_by_sector"], g[COL_OPM], frac=0.3)
            if len(xs) > 1:
                plt.plot(xs, ys, linewidth=2)
            plt.title(f"OPM % vs Net Profit (scaled) — {sec}")
            plt.xlabel("Net Profit (robust z-score within sector)")
            plt.ylabel("OPM %")
            plt.tight_layout()
            plt.show()

# ---------------- 3) Topline (Sales/Revenue) vs 3-Day Return (%) ----------------
work["Topline"] = work[COL_SALES].where(work[COL_SALES].notna(), work[COL_REV])
top = work[["Topline", COL_RET]].dropna()
top = top[top["Topline"] > 0]  # for log-scale

plt.figure(figsize=(7, 5))
plt.scatter(top["Topline"], top[COL_RET], s=18, alpha=0.6)
xs, ys = lowess_smooth(top["Topline"], top[COL_RET], frac=0.25)
if len(xs) > 1:
    plt.plot(xs, ys, linewidth=2)
plt.xscale("log")
plt.title("Topline (Sales/Revenue) vs 3-Day Return (%)")
plt.xlabel("Topline (log scale)")
plt.ylabel("3-Day Return (%)")
plt.tight_layout()
plt.show()

# ---------------- 4) BFSI: Financing Margin % vs Return ----------------
if COL_SECTOR in work.columns and COL_FM in work.columns:
    if work[COL_SECTOR].dtype == object:
        bfs = work[work[COL_SECTOR].str.upper() == "BFSI"]
    else:
        bfs = work[work[COL_SECTOR] == "BFSI"]
    bfs = bfs[[COL_FM, COL_RET]].dropna()
else:
    bfs = pd.DataFrame(columns=[COL_FM, COL_RET])

if len(bfs) >= 3:
    plt.figure(figsize=(7, 5))
    plt.scatter(bfs[COL_FM], bfs[COL_RET], s=20, alpha=0.7)
    xs, ys = lowess_smooth(bfs[COL_FM], bfs[COL_RET], frac=0.3)
    if len(xs) > 1:
        plt.plot(xs, ys, linewidth=2)
    plt.title("BFSI: Financing Margin % vs 3-Day Return (%)")
    plt.xlabel("Financing Margin %")
    plt.ylabel("3-Day Return (%)")
    plt.tight_layout()
    plt.show()
else:
    print("BFSI slice too small or lacks Financing Margin % data for a plot.")
```

```python
# ===== Cell — EDA: Outliers & influence (EPS, OPM %, Return) =====
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

assert "df" in globals(), "Please run the earlier cells that load `df` first."

# ---------------------------------------------
# Column names (exact, from your clean_data)
COL_EPS   = "EPS in Rs"
COL_OPM   = "OPM %"
COL_RET   = "3-Day Return (%)"
COL_TICK  = "Ticker"
COL_COMP  = "Company"
COL_SECT  = "Sector"
COL_CAP   = "Cap Category"
COL_RDATE = "Result Date"
COL_T     = "_T_date"
COL_T1    = "_T1_date"
COL_T2    = "_T2_date"
COL_T3    = "_T3_date"
COL_P_T   = "Closing Price (T)"
COL_P_T1  = "Closing Price (T+1)"
COL_P_T2  = "Closing Price (T+2)"
COL_P_T3  = "Closing Price (T+3)"
COL_MOV   = "Movement Label (1,0)"

work = df.copy()

# Ensure numeric for target features
for c in [COL_EPS, COL_OPM, COL_RET]:
    work[c] = pd.to_numeric(work[c], errors="coerce")

# Parse dates (dtype only; we are NOT altering the stored values)
for c in [COL_RDATE, COL_T, COL_T1, COL_T2, COL_T3]:
    if c in work.columns:
        work[c] = pd.to_datetime(work[c], errors="coerce")

# ---------- helper functions ----------
def iqr_bounds(s: pd.Series, k: float = 1.5):
    s = pd.to_numeric(s, errors="coerce").dropna()
    if s.empty:
        return (np.nan, np.nan)
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    return (q1 - k*iqr, q3 + k*iqr)

def z_score(s: pd.Series):
    s = pd.to_numeric(s, errors="coerce")
    mu = np.nanmean(s)
    sd = np.nanstd(s, ddof=0)
    if not np.isfinite(sd) or sd == 0:
        return pd.Series(np.nan, index=s.index)
    return (s - mu) / sd

def robust_z(s: pd.Series):
    s = pd.to_numeric(s, errors="coerce")
    med = np.nanmedian(s)
    mad = np.nanmedian(np.abs(s - med))
    if not np.isfinite(mad) or mad == 0:
        return pd.Series(np.nan, index=s.index)
    return 0.6745 * (s - med) / mad  # ~N(0,1) under normality

def non_monotonic_T_block(row):
    # Any violation of T < T1 < T2 < T3 (strictly increasing calendar order)
    t, t1, t2, t3 = row[COL_T], row[COL_T1], row[COL_T2], row[COL_T3]
    if pd.isna(t) or pd.isna(t1) or pd.isna(t2) or pd.isna(t3):
        return np.nan
    return not (t < t1 < t2 < t3)

def t_after_result(row):
    t, r = row[COL_T], row[COL_RDATE]
    if pd.isna(t) or pd.isna(r):
        return np.nan
    return bool(t > r)

# ---------- compute outlier flags ----------
metrics = { "EPS": COL_EPS, "OPM": COL_OPM, "RET": COL_RET }
summ_rows = []

for tag, col in metrics.items():
    # bounds for IQR
    lo, hi = iqr_bounds(work[col], k=1.5)
    work[f"{tag}_z"]  = z_score(work[col])
    work[f"{tag}_rz"] = robust_z(work[col])

    work[f"{tag}_is_outlier_IQR"] = (work[col] < lo) | (work[col] > hi)
    work[f"{tag}_is_outlier_Z3"]  = work[f"{tag}_z"].abs() >= 3
    work[f"{tag}_is_outlier_RZ35"] = work[f"{tag}_rz"].abs() >= 3.5

    summ_rows.append([tag, "IQR(1.5×)", int(work[f"{tag}_is_outlier_IQR"].sum()), f"lo={lo:.3f}, hi={hi:.3f}"])
    summ_rows.append([tag, "|Z|≥3",     int(work[f"{tag}_is_outlier_Z3"].sum()),  "μ±3σ"])
    summ_rows.append([tag, "|Robust Z|≥3.5", int(work[f"{tag}_is_outlier_RZ35"].sum()), "median±3.5×MAD(1/0.6745)"])

out_summ = pd.DataFrame(summ_rows, columns=["Metric", "Rule", "Outlier Count", "Rule Detail"])
print("\n=== Outlier Summary (counts by method) ===")
display(out_summ)

# ---------- top influence points (by robust z) ----------
def top_influential(col_name: str, tag: str, k: int = 12):
    rz = work[f"{tag}_rz"]
    take = work.loc[rz.abs().sort_values(ascending=False).head(k).index, [
        COL_COMP, COL_TICK, COL_SECT, COL_CAP, COL_RDATE,
        col_name, f"{tag}_rz", f"{tag}_z",
        COL_T, COL_T1, COL_T2, COL_T3, COL_P_T, COL_P_T1, COL_P_T2, COL_P_T3, COL_MOV
    ]].copy()

    # Influence notes: (1) T after Result?  (2) T dates non-monotonic?
    take["T_AFTER_RESULT"] = work.loc[take.index].apply(t_after_result, axis=1)
    take["NON_MONOTONIC_T_BLOCK"] = work.loc[take.index].apply(non_monotonic_T_block, axis=1)

    # A tiny, readable rename
    ren = {f"{tag}_rz": "Robust_Z", f"{tag}_z": "Z", col_name: "Value"}
    take = take.rename(columns=ren).sort_values("Robust_Z", key=lambda s: s.abs(), ascending=False)
    return take

print("\n=== Top influence (by |Robust Z|) — 3-Day Return (%) ===")
display(top_influential(COL_RET, "RET", k=15))

print("\n=== Top influence (by |Robust Z|) — EPS in Rs ===")
display(top_influential(COL_EPS, "EPS", k=12))

print("\n=== Top influence (by |Robust Z|) — OPM % ===")
display(top_influential(COL_OPM, "OPM", k=12))

# ---------- quick boxplots for visual sense (optional) ----------
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, col, title in zip(
    axes,
    [COL_RET, COL_EPS, COL_OPM],
    ["3-Day Return (%)", "EPS in Rs", "OPM %"]
):
    ax.boxplot(work[col].dropna(), vert=True, showfliers=True)
    ax.set_title(title)
plt.tight_layout()
plt.show()

# Small recap table
recap = pd.DataFrame({
    "Rows": [len(work)],
    "Non-null Return": [work[COL_RET].notna().sum()],
    "Non-null EPS": [work[COL_EPS].notna().sum()],
    "Non-null OPM": [work[COL_OPM].notna().sum()]
})
print("\n=== Coverage recap ===")
display(recap)
```

```python
# ===== Cell — EDA: Sector-specific panels (BFSI + IT/FMCG/Pharma/Auto) =====


# Try to import SciPy for tests; fall back gracefully if not installed
try:
    from scipy import stats
    _HAVE_SCIPY = True
except Exception:
    _HAVE_SCIPY = False
    print("Note: SciPy not found; skipping t-test / Mann-Whitney p-values. Install with: pip install scipy")

assert "df" in globals(), "Please run the earlier cells that load `df` first."

# ----- Exact column names from clean_data -----
COL_COMP = "Company"
COL_TICK = "Ticker"
COL_SECT = "Sector"
COL_CAP  = "Cap Category"

COL_SALES   = "Sales"
COL_REVENUE = "Revenue"
COL_EPS     = "EPS in Rs"
COL_OPM     = "OPM %"
COL_NP      = "Net Profit"
COL_FM      = "Financing Margin %"        # BFSI
COL_GNPA    = "Gross NPA %"               # may or may not exist
COL_NNPA    = "Net NPA %"                 # may or may not exist

COL_RET   = "3-Day Return (%)"
COL_LABEL = "Movement Label (1,0)"

# Ensure key numeric columns are numeric
for c in [COL_EPS, COL_OPM, COL_NP, COL_SALES, COL_REVENUE, COL_FM, COL_RET, COL_GNPA, COL_NNPA]:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors="coerce")

# Helper: choose "topline" (Sales or Revenue) without mutating df
def topline_series(frame: pd.DataFrame) -> pd.Series:
    s_sales = frame[COL_SALES] if COL_SALES in frame.columns else pd.Series(index=frame.index, dtype=float)
    s_rev   = frame[COL_REVENUE] if COL_REVENUE in frame.columns else pd.Series(index=frame.index, dtype=float)
    return s_sales.combine_first(s_rev)

# Helper: contrasts Up vs Down inside a sector for one feature
def sector_contrasts(sec_df: pd.DataFrame, feature: str) -> dict:
    out = {
        "Feature": feature,
        "Up_n": np.nan, "Down_n": np.nan,
        "Up_mean": np.nan, "Down_mean": np.nan,
        "Up_median": np.nan, "Down_median": np.nan,
        "Cohen_d": np.nan, "t_p": np.nan, "mw_p": np.nan
    }
    if feature not in sec_df.columns:
        return out

    s = pd.to_numeric(sec_df[feature], errors="coerce")
    m = pd.to_numeric(sec_df[COL_LABEL], errors="coerce")

    x = s[m == 1].dropna()
    y = s[m == 0].dropna()

    out.update({
        "Up_n": int(x.shape[0]),
        "Down_n": int(y.shape[0]),
        "Up_mean": float(np.nanmean(x)) if x.size else np.nan,
        "Down_mean": float(np.nanmean(y)) if y.size else np.nan,
        "Up_median": float(np.nanmedian(x)) if x.size else np.nan,
        "Down_median": float(np.nanmedian(y)) if y.size else np.nan,
    })

    # Cohen's d
    if x.size and y.size:
        vx, vy = np.nanvar(x, ddof=1), np.nanvar(y, ddof=1)
        nx, ny = x.size, y.size
        pooled = np.sqrt(((nx - 1)*vx + (ny - 1)*vy) / (nx + ny - 2)) if (nx + ny - 2) > 0 else np.nan
        d = (np.nanmean(x) - np.nanmean(y)) / pooled if (pooled and np.isfinite(pooled) and pooled != 0) else np.nan
        out["Cohen_d"] = float(d) if np.isfinite(d) else np.nan

        if _HAVE_SCIPY and nx > 1 and ny > 1:
            try:
                t_p = stats.ttest_ind(x, y, equal_var=False, nan_policy="omit").pvalue
            except Exception:
                t_p = np.nan
            try:
                mw_p = stats.mannwhitneyu(x, y, alternative="two-sided").pvalue
            except Exception:
                mw_p = np.nan
            out["t_p"] = float(t_p) if np.isfinite(t_p) else np.nan
            out["mw_p"] = float(mw_p) if np.isfinite(mw_p) else np.nan

    return out

# --------------------------
# 1) BFSI-only panel
# --------------------------
print("\n================= BFSI PANEL =================")
bfsi = df[df[COL_SECT] == "BFSI"].copy()
print(f"BFSI rows: {len(bfsi)} | tickers: {bfsi[COL_TICK].nunique()}")

# Distributions: Financing Margin %, GNPA, NNPA (if present)
figs = []
if COL_FM in bfsi.columns and bfsi[COL_FM].notna().any():
    plt.figure(figsize=(6,3.2))
    plt.hist(bfsi[COL_FM].dropna(), bins=20)
    plt.title("BFSI — Financing Margin % (distribution)")
    plt.xlabel("Financing Margin %")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()
else:
    print("Note: Financing Margin % not available or all missing in BFSI subset.")

for col, title in [(COL_GNPA, "Gross NPA %"), (COL_NNPA, "Net NPA %")]:
    if col in bfsi.columns and bfsi[col].notna().any():
        plt.figure(figsize=(6,3.2))
        plt.hist(bfsi[col].dropna(), bins=20)
        plt.title(f"BFSI — {title} (distribution)")
        plt.xlabel(title)
        plt.ylabel("Count")
        plt.tight_layout()
        plt.show()
    else:
        print(f"Note: {title} column not present or all missing; skipping.")

# Relation to return / labels: FM vs Return (scatter), box by Up/Down
if COL_FM in bfsi.columns and bfsi[COL_FM].notna().any() and bfsi[COL_RET].notna().any():
    # Scatter FM vs Return
    plt.figure(figsize=(6,4))
    m = pd.to_numeric(bfsi[COL_LABEL], errors="coerce").fillna(0).astype(int)
    up_mask = m == 1
    plt.scatter(bfsi.loc[~up_mask, COL_FM], bfsi.loc[~up_mask, COL_RET], s=18, alpha=0.7, label="Down (0)")
    plt.scatter(bfsi.loc[up_mask,  COL_FM], bfsi.loc[up_mask,  COL_RET], s=18, alpha=0.7, label="Up (1)")
    plt.xlabel("Financing Margin %")
    plt.ylabel("3-Day Return (%)")
    plt.title("BFSI — Financing Margin % vs 3-Day Return (%)")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Box FM by label
    up_vals = bfsi.loc[up_mask, COL_FM].dropna()
    dn_vals = bfsi.loc[~up_mask, COL_FM].dropna()
    if len(up_vals) and len(dn_vals):
        plt.figure(figsize=(5,3.5))
        plt.boxplot([dn_vals, up_vals], labels=["Down (0)", "Up (1)"], showfliers=True)
        plt.ylabel("Financing Margin %")
        plt.title("BFSI — Financing Margin % by Movement Label")
        plt.tight_layout()
        plt.show()

# BFSI contrasts table
bfsi_features = [c for c in [COL_FM, COL_GNPA, COL_NNPA] if c in df.columns]
if bfsi_features:
    rows = [sector_contrasts(bfsi, f) for f in bfsi_features]
    bfsi_contr = pd.DataFrame(rows)
    print("\nBFSI — Up/Down contrasts:")
    display(bfsi_contr)
else:
    print("No BFSI numeric fundamentals (FM/GNPA/NNPA) available to contrast.")

# --------------------------
# 2) Sector panels: IT / FMCG / Pharma / Auto
# --------------------------
print("\n=========== SECTOR PANELS (IT, FMCG, Pharma, Auto) ===========")

sector_feature_map = {
    "IT / Technology": [COL_OPM, COL_EPS, "Topline"],  # OPM %, EPS, Sales/Revenue
    "FMCG":            [COL_OPM, "Topline", COL_EPS],
    "Pharma":          [COL_OPM, COL_NP,   COL_EPS],
    "Auto":            [COL_OPM, "Topline", COL_EPS],
}

all_contrast_tables = []

for sector, feats in sector_feature_map.items():
    sec_df = df[df[COL_SECT] == sector].copy()
    if sec_df.empty:
        print(f"\n{sector}: no rows; skipping.")
        continue
    print(f"\n--- {sector} --- rows={len(sec_df)} | tickers={sec_df[COL_TICK].nunique()}")

    # Make a working copy with a temporary 'Topline' column (Sales or Revenue)
    sec_df = sec_df.copy()
    sec_df["Topline"] = topline_series(sec_df)

    # Visuals: hist of key features (available only)
    available_feats = [COL for COL in feats if (COL in sec_df.columns or COL == "Topline")]
    nplots = len(available_feats)
    if nplots:
        plt.figure(figsize=(5.5*nplots, 3.2))
        for i, col in enumerate(available_feats, 1):
            vals = pd.to_numeric(sec_df[col], errors="coerce").dropna()
            plt.subplot(1, nplots, i)
            if len(vals):
                plt.hist(vals, bins=20)
                plt.title(f"{sector} — {col} (dist)")
                plt.xlabel(col); plt.ylabel("Count")
            else:
                plt.title(f"{sector} — {col}: no data")
        plt.tight_layout()
        plt.show()

    # Up/Down contrasts per feature
    rows = []
    for col in available_feats:
        rows.append(sector_contrasts(sec_df, col))
    sec_contr = pd.DataFrame(rows)
    print(f"{sector} — Up/Down contrasts:")
    display(sec_contr)
    all_contrast_tables.append(sec_contr.assign(Sector=sector))

# Combined summary (stacked)
if all_contrast_tables:
    combined = pd.concat(all_contrast_tables, ignore_index=True)
    # Order columns nicely
    cols = ["Sector","Feature","Up_n","Down_n","Up_mean","Down_mean","Up_median","Down_median","Cohen_d","t_p","mw_p"]
    combined = combined.reindex(columns=cols)
    print("\nAll sectors — contrasts (stacked):")
    display(combined)
else:
    print("No sector contrasts to display (no data).")

# Small recap
recap = (
    df.groupby(COL_SECT)[COL_TICK]
      .agg(rows="count", tickers="nunique")
      .reset_index()
)
print("\nSector coverage recap:")
display(recap)
```

```python
#AA# ==== Cell — Synthesis tables & charts (insight-first) ====
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Optional deps
try:
    from scipy import stats
    HAVE_SCIPY = True
except Exception:
    HAVE_SCIPY = False
    print("Note: SciPy not found; significance tests will be skipped. Install: pip install scipy")

try:
    from sklearn.feature_selection import mutual_info_classif
    HAVE_SKLEARN = True
except Exception:
    HAVE_SKLEARN = False
    print("Note: scikit-learn not found; mutual information will be skipped. Install: pip install scikit-learn")

assert "df" in globals(), "Please run the earlier cells that load `df` first."

# ----- Column names (exact) -----
COL_SECTOR   = "Sector"
COL_CAP      = "Cap Category"
COL_RETURN   = "3-Day Return (%)"
COL_LABEL    = "Movement Label (1,0)"

PRICE_COLS = [
    "Closing Price (T)", "Closing Price (T+1)",
    "Closing Price (T+2)", "Closing Price (T+3)",
    "3-Day Avg Price Post Result"
]

# Candidate numeric fundamentals (keep only those that actually exist)
CAND_NUM = [
    "Sales","Expenses","Operating Profit","OPM %","Other Income","Interest",
    "Depreciation","Profit before tax","Tax %","Net Profit","EPS in Rs",
    "Revenue","Financing Profit","Financing Margin %"
] + PRICE_COLS + [COL_RETURN]

NUM_EXIST = [c for c in CAND_NUM if c in df.columns]

# Ensure numeric dtype
for c in NUM_EXIST:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Features to correlate with return (exclude direct price-derived leakage)
FEATURES_FOR_CORR = [c for c in NUM_EXIST if c not in (PRICE_COLS + [COL_RETURN])]

print("Correlation features (excluding price-derived leakage):")
print(FEATURES_FOR_CORR)

# ---------- 1) Top 10 correlation with 3-Day Return ----------
def corr_with_target(frame, cols, target, method="pearson"):
    out = []
    y = pd.to_numeric(frame[target], errors="coerce")
    for c in cols:
        x = pd.to_numeric(frame[c], errors="coerce")
        pair = pd.concat([x, y], axis=1).dropna()
        if pair.shape[0] < 5:
            r = np.nan
        else:
            if method == "spearman":
                r = pair.corr(method="spearman").iloc[0,1]
            else:
                r = pair.corr(method="pearson").iloc[0,1]
        out.append({"Feature": c, "n": int(pair.shape[0]), "r": r})
    res = pd.DataFrame(out).sort_values("r", key=lambda s: s.abs(), ascending=False)
    return res

corr_p = corr_with_target(df, FEATURES_FOR_CORR, COL_RETURN, method="pearson").head(10)
corr_s = corr_with_target(df, FEATURES_FOR_CORR, COL_RETURN, method="spearman").head(10)

print("\nTop 10 features by Pearson correlation with 3-Day Return (%):")
display(corr_p)

print("\nTop 10 features by Spearman correlation with 3-Day Return (%):")
display(corr_s)

# Quick bars for Pearson
plt.figure(figsize=(8,4))
plt.barh(corr_p["Feature"][::-1], corr_p["r"][::-1])
plt.axvline(0, linewidth=1)
plt.title("Top 10 (Pearson) correlation with 3-Day Return (%)")
plt.tight_layout()
plt.show()

# ---------- 2) Top 10 mutual information with Movement Label ----------
if HAVE_SKLEARN and COL_LABEL in df.columns:
    X = df[FEATURES_FOR_CORR].copy()
    # Impute missing with column medians for MI
    X = X.apply(lambda s: pd.to_numeric(s, errors="coerce"))
    X = X.fillna(X.median(numeric_only=True))
    y = pd.to_numeric(df[COL_LABEL], errors="coerce").fillna(0).astype(int)

    try:
        mi = mutual_info_classif(X.values, y.values, discrete_features=False, random_state=42)
        mi_df = pd.DataFrame({"Feature": X.columns, "MI": mi}).sort_values("MI", ascending=False).head(10)
        print("\nTop 10 features by mutual information with Movement Label:")
        display(mi_df)

        # Bars
        plt.figure(figsize=(8,4))
        plt.barh(mi_df["Feature"][::-1], mi_df["MI"][::-1])
        plt.title("Top 10 Mutual Information with Movement Label")
        plt.tight_layout()
        plt.show()
    except Exception as e:
        print(f"MI computation error: {e}")
else:
    print("\nMutual information skipped (scikit-learn missing or Movement Label absent).")

# ---------- 3) Sectors ranked by average 3-Day Return ----------
if COL_SECTOR in df.columns:
    sector_rank = (
        df[[COL_SECTOR, COL_RETURN]]
        .dropna(subset=[COL_RETURN])
        .groupby(COL_SECTOR)
        .agg(n=("3-Day Return (%)","count"),
             mean_ret=("3-Day Return (%)","mean"),
             median_ret=("3-Day Return (%)","median"))
        .sort_values("mean_ret", ascending=False)
        .reset_index()
    )
    print("\nSectors ranked by average 3-Day Return (%):")
    display(sector_rank)

    # Chart
    plt.figure(figsize=(8,4))
    plt.barh(sector_rank[COL_SECTOR][::-1], sector_rank["mean_ret"][::-1])
    plt.title("Average 3-Day Return (%) by Sector")
    plt.tight_layout()
    plt.show()
else:
    print("\nSector column not found; skipping sector ranking.")

# ---------- 4) Cap category summary ----------
if COL_CAP in df.columns:
    cap_summary = (
        df[[COL_CAP, COL_RETURN, COL_LABEL]]
        .dropna(subset=[COL_LABEL])
        .assign(_is_up=lambda t: (pd.to_numeric(t[COL_LABEL], errors="coerce") == 1).astype(int))
        .groupby(COL_CAP)
        .agg(count=(" _is_up","count") if " _is_up" in df.columns else ("_is_up","count"),
             mean_return=(COL_RETURN,"mean"),
             up_pct=("_is_up","mean"))
        .reset_index()
    )
    # Fix possible misnamed column due to groupby tuple quirk
    if "count" not in cap_summary.columns:
        cap_summary = cap_summary.rename(columns={cap_summary.columns[1]: "count"})
    cap_summary["up_pct"] = cap_summary["up_pct"] * 100.0

    print("\nCap category summary (counts, mean return, Up%):")
    display(cap_summary)

    # Chart
    plt.figure(figsize=(6,3.6))
    plt.bar(cap_summary[COL_CAP], cap_summary["up_pct"])
    plt.title("Up% by Cap Category")
    plt.ylabel("Up (%)")
    plt.tight_layout()
    plt.show()
else:
    print("\nCap Category column not found; skipping cap summary.")

# ---------- 5) Statistically significant Up vs Down differences ----------
def cohen_d(x, y):
    x = np.asarray(x, dtype=float); y = np.asarray(y, dtype=float)
    if x.size < 2 or y.size < 2:
        return np.nan
    vx, vy = np.nanvar(x, ddof=1), np.nanvar(y, ddof=1)
    nx, ny = x.size, y.size
    denom = np.sqrt(((nx - 1)*vx + (ny - 1)*vy) / (nx + ny - 2)) if (nx + ny - 2) > 0 else np.nan
    if not np.isfinite(denom) or denom == 0:
        return np.nan
    return (np.nanmean(x) - np.nanmean(y)) / denom

NUM_FOR_TESTS = [c for c in NUM_EXIST if c not in ([COL_RETURN] + PRICE_COLS)]  # exclude target and price leakage
res_rows = []
if HAVE_SCIPY and COL_LABEL in df.columns:
    lab = pd.to_numeric(df[COL_LABEL], errors="coerce")
    for feat in NUM_FOR_TESTS:
        vals = pd.to_numeric(df[feat], errors="coerce")
        pair = pd.concat([vals, lab], axis=1).dropna()
        if pair.shape[0] < 8:
            continue
        x = pair[pair[COL_LABEL] == 1][feat].values
        y = pair[pair[COL_LABEL] == 0][feat].values
        if x.size < 3 or y.size < 3:
            continue
        try:
            t_p = stats.ttest_ind(x, y, equal_var=False, nan_policy="omit").pvalue
        except Exception:
            t_p = np.nan
        try:
            mw_p = stats.mannwhitneyu(x, y, alternative="two-sided").pvalue
        except Exception:
            mw_p = np.nan
        d = cohen_d(x, y)
        best_p = np.nanmin([t_p, mw_p])
        res_rows.append({
            "Feature": feat,
            "Up_n": int(x.size), "Down_n": int(y.size),
            "Cohen_d": d,
            "t_p": t_p, "mw_p": mw_p, "Best_p": best_p
        })
    signif = (
        pd.DataFrame(res_rows)
        .dropna(subset=["Best_p"])
        .sort_values("Best_p")
    )
    # Threshold (no multiple-comparison correction here; we’ll just show raw p)
    signif_05 = signif[signif["Best_p"] < 0.05].copy()
    print("\nStatistically significant Up vs Down differences (any test p<0.05):")
    display(signif_05.head(20) if not signif_05.empty else pd.DataFrame(columns=["Feature","Up_n","Down_n","Cohen_d","t_p","mw_p","Best_p"]))
else:
    print("\nSignificance tests skipped (SciPy missing or label column absent).")

print("\n—— Synthesis complete.")
```
